## 1. Headline finding

5 Paraguayan Verra projects systematically under-claim carbon loss by ~35%.
Total Verra claim: 3.30 Mt CO₂e.  
Hansen-derived estimate: 4.44 Mt CO₂e.  
Discrepancy: +1.14 Mt (+34.6%)

```python
import sys
sys.path.insert(0, '../../')
import numpy as np
import matplotlib.pyplot as plt
import json
from pathlib import Path

REPO = Path('../..').resolve()

verra_file = REPO / 'outputs/carbon_credits/verra_verification.json'
data = json.loads(verra_file.read_text())
projects = data['projects']

names = [p['name'] for p in projects]
verra = [p['verra_claim_mt'] for p in projects]
hansen = [p['hansen_estimate_mt'] for p in projects]

fig, ax = plt.subplots(figsize=(10, 6))
x = np.arange(len(names))
w = 0.35
ax.bar(x - w/2, verra, w, label='Verra claim', color='#1f77b4')
ax.bar(x + w/2, hansen, w, label='Hansen estimate', color='#ff7f0e')
ax.set_xticks(x)
ax.set_xticklabels(names, rotation=20, ha='right')
ax.set_ylabel('CO₂e (Mt)')
ax.set_title('Verra vs Hansen estimates (5 Paraguayan projects)')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2. Discrepancy analysis

All 5 projects show positive discrepancy (Verra under-claims).

```python
discrepancies = [p['discrepancy_pct'] for p in projects]
print(f'Discrepancies: {discrepancies}')
print(f'Mean: {np.mean(discrepancies):.1f}%')
print(f'Std: {np.std(discrepancies):.1f}%')

# Statistical test: are these systematically biased?
from scipy.stats import ttest_1samp
t_stat, p_value = ttest_1samp(discrepancies, 0)
print(f'One-sample t-test against 0: t={t_stat:.2f}, p={p_value:.4f}')
print(f'Verra under-claims are statistically significant (p<0.001)' if p_value < 0.001 else 'Not significant')
```
