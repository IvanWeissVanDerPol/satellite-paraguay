````python
# Verify a sample project
if len(verra) > 0:
    project_id = verra.iloc[0]['id']
    result = pipeline.verify_carbon_credit(
        project_id=project_id,
        tile_id='-54.267_-21.164',
    )
    print(f"\nVerification result:")
    print(f"  Project: {result['project_id']}")
    print(f"  Claimed: {result['claimed_carbon_tons']} tons")
    print(f"  Estimated: {result['estimated_carbon_tons']} tons")
    print(f"  Confidence: {result['confidence']:.2f}")
    print(f"  Verified: {result['verified']}")
````
